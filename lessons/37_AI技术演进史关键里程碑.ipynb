## 核心概念：AI 的五次浪潮

AI 不是线性进步，而是一波三折的浪潮式演进。每一波浪潮都由**三大要素**驱动：

| 要素 | 含义 | 类比 |
|------|------|------|
| **算法** | 计算方法与模型架构 | 发动机设计 |
| **算力** | 硬件计算能力 | 发动机功率 |
| **数据** | 训练与验证用数据集 | 燃料 |

当三个要素同时就位时，AI 就会爆发；缺一个就会陷入寒冬。

```
浪潮1: 符号主义 (1956-1974) → 第一次寒冬
浪潮2: 专家系统 (1980-1987) → 第二次寒冬  
浪潮3: 统计学习 (1995-2011) → 渐进积累
浪潮4: 深度学习 (2012-2020) → 产业化
浪潮5: 大模型时代 (2020-至今) → 通用化
```

````python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# AI 历史浪潮可视化
fig, ax = plt.subplots(figsize=(14, 6))

# 定义浪潮时间线
waves = [
    {'name': '符号主义', 'start': 1956, 'end': 1974, 'color': '#4A90D9', 'peak': 1970},
    {'name': '专家系统', 'start': 1980, 'end': 1987, 'color': '#7B68EE', 'peak': 1985},
    {'name': '统计学习', 'start': 1995, 'end': 2011, 'color': '#2ECC71', 'peak': 2005},
    {'name': '深度学习', 'start': 2012, 'end': 2020, 'color': '#E74C3C', 'peak': 2017},
    {'name': '大模型时代', 'start': 2020, 'end': 2026, 'color': '#C96442', 'peak': 2024},
]

for i, w in enumerate(waves):
    x = np.linspace(w['start'], w['end'], 100)
    # 用高斯曲线模拟热度
    peak_year = (w['peak'] - w['start']) / (w['end'] - w['start'])
    y = np.exp(-((np.linspace(0, 1, 100) - peak_year) ** 2) / 0.08) * (1 + i * 0.3)
    ax.fill_between(x, 0, y, alpha=0.3, color=w['color'])
    ax.plot(x, y, color=w['color'], linewidth=2)
    ax.text(w['peak'], max(y) + 0.1, w['name'], ha='center', fontsize=10, fontweight='bold', color=w['color'])

# 标注关键事件
events = [
    (1956, '达特茅斯\n会议'),
    (2012, 'AlexNet\n突破'),
    (2017, 'Transformer\n论文'),
    (2022, 'ChatGPT\n发布'),
]
for year, label in events:
    ax.axvline(x=year, color='gray', linestyle='--', alpha=0.5)
    ax.text(year, 2.8, label, ha='center', fontsize=8, color='#333',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='#FAF8F5', edgecolor='#ccc'))

ax.set_xlim(1950, 2027)
ax.set_ylim(0, 3.2)
ax.set_xlabel('年份', fontsize=12)
ax.set_ylabel('AI 热度/影响力', fontsize=12)
ax.set_title('AI 技术演进浪潮图', fontsize=14, fontweight='bold')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.savefig('ai_history_waves.png', dpi=150, bbox_inches='tight')
plt.show()
print('✅ AI 浪潮图生成完成')
````

## 总结

### AI 演进的 5 个关键洞察

| # | 洞察 | 含义 |
|---|------|------|
| 1 | **算法-算力-数据三角** | 每次突破都三者齐备，缺一则停滞 |
| 2 | **简单方法先有效** | 线性回归→决策树→神经网络，复杂度递增有原因 |
| 3 | **Transformer 是分水岭** | 2017 年之后，几乎所有突破都基于 Transformer |
| 4 | **规模是当前引擎** | 更多参数+更多数据+更多算力 → 更好能力 |
| 5 | **工程化决定落地** | 算法突破后，工程化能力决定谁能用上 |
